```python
# Cell 16 — Numeric constraint helper

def apply_max_constraint(
    dataframe,
    field,
    maximum
):
    """
    Applies a maximum-value constraint.

    Rules:
    1. Known value <= maximum  → PASS
    2. Known value > maximum   → REJECT
    3. Missing value            → KEEP for later validation

    Missing dimensions are not automatically rejected because
    they may be recoverable or may not have been specified
    in the product source.
    """

    if maximum is None:
        return dataframe

    if field not in dataframe.columns:
        return dataframe

    values = pd.to_numeric(
        dataframe[field],
        errors="coerce"
    )

    # Keep products when:
    # - the value is missing, OR
    # - the value satisfies the maximum constraint
    mask = values.isna() | (values <= maximum)

    return dataframe[mask]
```

```python
# Cell 17 — Structured product retrieval

def filter_products(
    dataframe,
    category=None,
    max_width_mm=None,
    max_depth_mm=None,
    max_height_mm=None
):
    """
    Structured retrieval using hard product constraints.
    
    Parameters:
        category:
            Normalized product category.
        
        max_width_mm:
            Maximum acceptable product width.
        
        max_depth_mm:
            Maximum acceptable product depth.
        
        max_height_mm:
            Maximum acceptable product height.
    """

    result = dataframe.copy()

    # -----------------------------
    # Category filter
    # -----------------------------
    if category is not None:
        result = result[
            result["category_normalized"]
            .fillna("Other")
            .str.lower()
            == category.strip().lower()
        ]

    # -----------------------------
    # Width constraint
    # -----------------------------
    result = apply_max_constraint(
        result,
        "dimensions.width_mm",
        max_width_mm
    )

    # -----------------------------
    # Depth constraint
    # -----------------------------
    result = apply_max_constraint(
        result,
        "dimensions.depth_mm",
        max_depth_mm
    )

    # -----------------------------
    # Height constraint
    # -----------------------------
    result = apply_max_constraint(
        result,
        "dimensions.height_mm",
        max_height_mm
    )

    return result.reset_index(drop=True)
```

```python
# Cell 18 — Installation type filter

def apply_installation_constraint(
    dataframe,
    installation_type=None
):
    """
    Filters products by installation type.

    Products with missing installation information are
    kept as candidates and can be validated later.
    """

    if installation_type is None:
        return dataframe

    if "installation.type" not in dataframe.columns:
        return dataframe

    values = (
        dataframe["installation.type"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str.lower()
    )

    requested = installation_type.strip().lower()

    # Exact match OR unknown installation
    mask = (
        values.eq(requested)
        | values.eq("")
        | values.eq("not specified")
    )

    return dataframe[mask]
```

```python
# Cell 19 — Rough-in constraint

def apply_rough_in_constraint(
    dataframe,
    required_rough_in_mm=None
):
    """
    Filters products based on rough-in.

    If rough-in is available, it must match the requested
    rough-in value.

    Products with unavailable rough-in are retained for
    later validation.
    """

    if required_rough_in_mm is None:
        return dataframe

    field = "installation.rough_in_mm"

    if field not in dataframe.columns:
        return dataframe

    values = pd.to_numeric(
        dataframe[field],
        errors="coerce"
    )

    # Allow a small tolerance because installation measurements
    # may have practical variation.
    tolerance = 10

    mask = (
        values.isna()
        | (
            (values - required_rough_in_mm).abs()
            <= tolerance
        )
    )

    return dataframe[mask]
```

```python
# Cell 20 — Electrical requirement filter

def apply_electrical_constraint(
    dataframe,
    electrical_required=None
):
    """
    Filters products based on electrical requirements.

    If electrical requirement is unknown, the product is
    retained for later validation.
    """

    if electrical_required is None:
        return dataframe

    field = "electrical.required"

    if field not in dataframe.columns:
        return dataframe

    values = dataframe[field]

    # Convert common textual values to boolean
    normalized = (
        values.astype(str)
        .str.strip()
        .str.lower()
    )

    known_true = normalized.isin(
        ["true", "yes", "1"]
    )

    known_false = normalized.isin(
        ["false", "no", "0"]
    )

    # Unknown values
    unknown = ~(known_true | known_false)

    if electrical_required:
        mask = known_true | unknown
    else:
        mask = known_false | unknown

    return dataframe[mask]
```

```python
# Cell 21 — Complete structured retrieval function

def filter_products(
    dataframe,
    category=None,
    max_width_mm=None,
    max_depth_mm=None,
    max_height_mm=None,
    installation_type=None,
    required_rough_in_mm=None,
    electrical_required=None
):
    """
    Main structured retrieval function.

    Applies hard product constraints using structured
    product metadata.
    """

    result = dataframe.copy()

    # -----------------------------
    # Category
    # -----------------------------
    if category is not None:
        result = result[
            result["category_normalized"]
            .fillna("Other")
            .str.lower()
            == category.strip().lower()
        ]

    # -----------------------------
    # Dimensions
    # -----------------------------
    result = apply_max_constraint(
        result,
        "dimensions.width_mm",
        max_width_mm
    )

    result = apply_max_constraint(
        result,
        "dimensions.depth_mm",
        max_depth_mm
    )

    result = apply_max_constraint(
        result,
        "dimensions.height_mm",
        max_height_mm
    )

    # -----------------------------
    # Installation
    # -----------------------------
    result = apply_installation_constraint(
        result,
        installation_type
    )

    # -----------------------------
    # Rough-in
    # -----------------------------
    result = apply_rough_in_constraint(
        result,
        required_rough_in_mm
    )

    # -----------------------------
    # Electrical
    # -----------------------------
    result = apply_electrical_constraint(
        result,
        electrical_required
    )

    return result.reset_index(drop=True)
```

```python
# Cell 24 — Search raw JSON recursively for price-related keys

def find_keys_containing(obj, keywords, path=""):
    """
    Recursively searches a JSON object for keys containing
    any of the supplied keywords.
    """

    found = []

    if isinstance(obj, dict):
        for key, value in obj.items():

            current_path = (
                f"{path}.{key}"
                if path
                else key
            )

            if any(
                keyword in key.lower()
                for keyword in keywords
            ):
                found.append(
                    (current_path, value)
                )

            found.extend(
                find_keys_containing(
                    value,
                    keywords,
                    current_path
                )
            )

    elif isinstance(obj, list):

        for index, item in enumerate(obj):

            current_path = f"{path}[{index}]"

            found.extend(
                find_keys_containing(
                    item,
                    keywords,
                    current_path
                )
            )

    return found


price_keys = find_keys_containing(
    products,
    ["price", "cost", "amount", "mrp"]
)

if price_keys:

    print("Price-related keys found:\n")

    for path, value in price_keys[:30]:
        print(path, "→", value)

else:
    print("No price-related keys found in the JSON.")
```

```python
# Cell 26 — Budget constraint

def apply_budget_constraint(
    dataframe,
    max_price_inr=None
):
    """
    Filters products using maximum budget.

    Known prices above the budget are rejected.

    Products without a known price are retained and
    marked for later price validation.
    """

    if max_price_inr is None:
        return dataframe

    if "price_inr" not in dataframe.columns:
        return dataframe

    prices = pd.to_numeric(
        dataframe["price_inr"],
        errors="coerce"
    )

    # Keep:
    # 1. Products with known price <= budget
    # 2. Products whose price is unknown
    mask = (
        prices.isna()
        | (prices <= max_price_inr)
    )

    return dataframe[mask]
```

```python
# Cell 27 — Updated structured retrieval function

def filter_products(
    dataframe,
    category=None,
    max_price_inr=None,
    max_width_mm=None,
    max_depth_mm=None,
    max_height_mm=None,
    installation_type=None,
    required_rough_in_mm=None,
    electrical_required=None
):
    """
    Main structured retrieval function.

    Uses exact product metadata for hard constraints.
    """

    result = dataframe.copy()

    # -----------------------------
    # Category
    # -----------------------------
    if category is not None:
        result = result[
            result["category_normalized"]
            .fillna("Other")
            .str.lower()
            == category.strip().lower()
        ]

    # -----------------------------
    # Budget
    # -----------------------------
    result = apply_budget_constraint(
        result,
        max_price_inr
    )

    # -----------------------------
    # Dimensions
    # -----------------------------
    result = apply_max_constraint(
        result,
        "dimensions.width_mm",
        max_width_mm
    )

    result = apply_max_constraint(
        result,
        "dimensions.depth_mm",
        max_depth_mm
    )

    result = apply_max_constraint(
        result,
        "dimensions.height_mm",
        max_height_mm
    )

    # -----------------------------
    # Installation
    # -----------------------------
    result = apply_installation_constraint(
        result,
        installation_type
    )

    # -----------------------------
    # Rough-in
    # -----------------------------
    result = apply_rough_in_constraint(
        result,
        required_rough_in_mm
    )

    # -----------------------------
    # Electrical
    # -----------------------------
    result = apply_electrical_constraint(
        result,
        electrical_required
    )

    return result.reset_index(drop=True)
```

```python
# Cell 28 — Constraint status helper

def check_max_value(value, maximum):
    """
    Checks whether a value satisfies a maximum constraint.

    Returns:
        True   -> constraint passed
        False  -> constraint failed
        None   -> value unavailable
    """

    if maximum is None:
        return True

    if pd.isna(value):
        return None

    return float(value) <= float(maximum)


def check_rough_in(value, required):
    """
    Checks rough-in compatibility with a 10 mm tolerance.
    """

    if required is None:
        return True

    if pd.isna(value):
        return None

    return abs(float(value) - float(required)) <= 10
```

```python
# Cell 29 — Validate one product against requirements

def validate_product(
    product,
    max_price_inr=None,
    max_width_mm=None,
    max_depth_mm=None,
    max_height_mm=None,
    required_rough_in_mm=None,
    installation_type=None,
    electrical_required=None
):
    """
    Produces a detailed validation report for one product.
    """

    checks = {}

    # -----------------------------
    # Price
    # -----------------------------
    if max_price_inr is not None:
        price = product.get("price_inr")

        if pd.isna(price):
            checks["budget"] = None
        else:
            checks["budget"] = float(price) <= max_price_inr

    # -----------------------------
    # Width
    # -----------------------------
    checks["width"] = check_max_value(
        product.get("dimensions.width_mm"),
        max_width_mm
    )

    # -----------------------------
    # Depth
    # -----------------------------
    checks["depth"] = check_max_value(
        product.get("dimensions.depth_mm"),
        max_depth_mm
    )

    # -----------------------------
    # Height
    # -----------------------------
    checks["height"] = check_max_value(
        product.get("dimensions.height_mm"),
        max_height_mm
    )

    # -----------------------------
    # Rough-in
    # -----------------------------
    checks["rough_in"] = check_rough_in(
        product.get("installation.rough_in_mm"),
        required_rough_in_mm
    )

    # -----------------------------
    # Installation
    # -----------------------------
    if installation_type is not None:

        actual = str(
            product.get("installation.type", "")
        ).strip().lower()

        requested = installation_type.strip().lower()

        if actual in ["", "not specified", "none"]:
            checks["installation"] = None
        else:
            checks["installation"] = (
                actual == requested
            )

    # -----------------------------
    # Electrical
    # -----------------------------
    if electrical_required is not None:

        value = product.get(
            "electrical.required"
        )

        if pd.isna(value):
            checks["electrical"] = None
        else:
            checks["electrical"] = (
                bool(value)
                == bool(electrical_required)
            )

    return checks
```

```python
# Cell 30 — Validation summary

def summarize_checks(checks):
    """
    Converts validation results into:
    
    PASS
    FAIL
    UNKNOWN
    """

    summary = {}

    for constraint, result in checks.items():

        if result is True:
            summary[constraint] = "PASS"

        elif result is False:
            summary[constraint] = "FAIL"

        else:
            summary[constraint] = "UNKNOWN"

    return summary
```

```python
# Cell 32 — Validation report for the catalog

def create_validation_report(
    dataframe,
    **requirements
):
    """
    Creates a validation report for every product.
    """

    reports = []

    for _, product in dataframe.iterrows():

        checks = validate_product(
            product,
            **requirements
        )

        summary = summarize_checks(checks)

        report = {
            "product_id": product.get("product_id"),
            "product_name": product.get("product_name"),
            "category": product.get("category_normalized"),
        }

        # Add validation statuses
        for constraint, status in summary.items():
            report[f"{constraint}_status"] = status

        reports.append(report)

    return pd.DataFrame(reports)
```

```python
def clean_metadata_value(value):
    """
    Convert values into formats that ChromaDB can safely
    store as metadata.
    """

    # Handle None
    if value is None:
        return "Not specified"

    # Handle pandas missing values
    if pd.isna(value):
        return "Not specified"

    # Handle lists such as features and colors
    if isinstance(value, list):
        return ", ".join(str(x) for x in value)

    return value


# Prepare data for ChromaDB

chroma_ids = []
chroma_documents = []
chroma_metadatas = []

for _, product in retrieval_df.iterrows():

    product_id = str(product["product_id"])

    metadata = {
        "product_id": product_id,

        "product_name": clean_metadata_value(
            product.get("product_name")
        ),

        "category": clean_metadata_value(
            product.get("category_normalized")
        ),

        "collection": clean_metadata_value(
            product.get("collection")
        ),

        "material": clean_metadata_value(
            product.get("material")
        ),

        "color": clean_metadata_value(
            product.get("color")
        ),

        "installation_type": clean_metadata_value(
            product.get("installation.type")
        )
    }

    chroma_ids.append(product_id)

    chroma_documents.append(
        product["semantic_text"]
    )

    chroma_metadatas.append(metadata)


print("ChromaDB data prepared successfully.")
print("IDs:", len(chroma_ids))
print("Documents:", len(chroma_documents))
print("Metadata:", len(chroma_metadatas))
```

```python
collection.upsert(
    ids=chroma_ids,
    documents=chroma_documents,
    metadatas=chroma_metadatas
)

print("Products added to ChromaDB.")
print("Number of records:", collection.count())
```
